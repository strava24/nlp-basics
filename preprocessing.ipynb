_nltk stands for Natural Language Toolkit._


In simple terms, it is a Python library that gives you tools for working with human language, such as English sentences, words, and paragraphs.

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
# Tokenization proces

nltk.download('punkt_tab') # a package that we need to download for tokenization
sentences = nltk.sent_tokenize(paragraph) # converting the pra into sentences and storing in a list
```

```python
sentences[0]
```

```python
stemmer = PorterStemmer()
stemmer.stem('history') # this is how a word is stemmed
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

nltk.download('wordnet')
```

```python
lemmatizer.lemmatize('history')
```

As we can see, lemmatizer is giving a meaningful stem, while the stemmer is giving a meaningless stem.. which is the main difference between them

```python
print(stemmer.stem("goes"))
print(lemmatizer.lemmatize("goes"))
```

```python
import re # python's regex module
```

```python
corpus = []

for i in range(len(sentences)):
    # replace every character that is not a letter (punctuation, digits, symbols) with a space
    review = re.sub('[^a-zA-Z]', ' ', sentences[i])
    # lowercase so that 'Luffy' and 'luffy' are treated as the same word
    review = review.lower()
    # split the sentence into a list of words
    review = review.split()
    # drop stopwords and lemmatize the remaining words (e.g. 'pirates' -> 'pirate')
    review = [lemmatizer.lemmatize(word) for word in review if word not in stopwords.words('english')]
    # join the words back into a single string, which is what CountVectorizer expects
    review = ' '.join(review)
    # store the cleaned sentence
    corpus.append(review)
```

```python
nltk.download('stopwords')

# Apply stemming
for i in corpus:

    words = nltk.word_tokenize(i) # tokenizing the sentences

    for word in words:
        if word not in stopwords.words('english'):
            print(stemmer.stem(word))
```

In simple terms, this is how bag of words work.

1. Collect every unique word in the corpus. This list is the **vocabulary**, and each word gets a fixed position (index) in it.
2. For each sentence, make a row of numbers with one slot per vocabulary word.
3. Fill each slot by looking at the sentence:
   - **Default bag of words:** the number of times the word appears (so `luffy` appearing twice gives `2`).
   - **Binary bag of words:** just `1` if the word appears and `0` if it doesn't.

The result is a matrix with one row per sentence and one column per word. The order of the words in a sentence is thrown away, as if we dropped all its words into a bag and only counted them. That is why it is called a "bag" of words.
